```python
import folium
import xyzservices.providers as xyz

# Primary corporate headquarters / campuses
companies = {
    "NVDA": {
        "name": "NVIDIA Headquarters",
        "location": [37.3706, -121.9672],  # Santa Clara, CA
        "color": "green"
    },
    "GOOGL": {
        "name": "Google (Alphabet) — Googleplex",
        "location": [37.4220, -122.0841],  # Mountain View, CA
        "color": "blue"
    },
    "MSFT": {
        "name": "Microsoft Headquarters",
        "location": [47.6396, -122.1280],  # Redmond, WA
        "color": "blue"
    },
    "AMZN": {
        "name": "Amazon Headquarters",
        "location": [47.6225, -122.3365],  # Seattle, WA
        "color": "orange"
    },
    "TSLA": {
        "name": "Tesla Headquarters",
        "location": [30.2220, -97.6170],   # Austin, TX
        "color": "red"
    },
    "META": {
        "name": "Meta Headquarters",
        "location": [37.4848, -122.1484],  # Menlo Park, CA
        "color": "blue"
    },
    "AAPL": {
        "name": "Apple Park",
        "location": [37.3349, -122.0090],  # Cupertino, CA
        "color": "gray"
    },
    "NFLX": {
        "name": "Netflix Headquarters",
        "location": [37.2599, -121.9636],  # Los Gatos, CA
        "color": "red"
    }
}

gray = xyz.Esri.WorldGrayCanvas

m = folium.Map(
    location=[39.5, -110.0],
    zoom_start=4,
    tiles=gray.build_url(),
    attr=gray.html_attribution
)


for ticker, company in companies.items():
    folium.Marker(
        location=company["location"],
        tooltip=f"{ticker} — {company['name']}",
        popup=folium.Popup(
            f"<b>{ticker}</b><br>{company['name']}",
            max_width=300
        ),
        icon=folium.Icon(
            color=company["color"],
            icon="info-sign"
        )
    ).add_to(m)


locations = [company["location"] for company in companies.values()]
m.fit_bounds(locations)

m
```

```python
ticker_list = ["NVDA", "GOOGL", "MSFT", "AMZN", "TSLA", "META", "AAPL", "NFLX"]
tickers = yf.Tickers(ticker_list)
info_list = [tickers.tickers[ticker].info for ticker in ticker_list]
df = pd.DataFrame(info_list)
company_names = [info.get("longName", "N/A") for info in info_list]
df.insert(0, "Company Name", company_names)
df.insert(1, "Ticker", ticker_list)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
df
```

```python
aapl_income = aapl.income_stmt
nvda_income = nvda.income_stmt
googl_income = googl.income_stmt
msft_income = msft.income_stmt
amzn_income = amzn.income_stmt
tsla_income = tsla.income_stmt
meta_income = meta.income_stmt
neflx_income = neflx.income_stmt

aapl_income = aapl_income.add_prefix("AAPL_")
nvda_income = nvda_income.add_prefix("NVDA_")
googl_income = googl_income.add_prefix("GOOGL_")
msft_income = msft_income.add_prefix("MSFT_")
amzn_income = amzn_income.add_prefix("AMZN_")
tsla_income = tsla_income.add_prefix("TSLA_")
meta_income = meta_income.add_prefix("META_")
neflx_income = neflx_income.add_prefix("NFLX_")

combined_df = pd.concat([aapl_income, nvda_income, googl_income, msft_income, amzn_income, tsla_income, meta_income, neflx_income], axis=1)
combined_df = combined_df.round().astype('Int64')

combined_df
combined_df.fillna(0).style.format("${:,.0f}")
```
